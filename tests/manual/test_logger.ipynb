# 日志模块演示
使用项目内的正式日志实现；输出目录由 BVC_DATA_DIR 控制，默认 var/logs。


```python
from bottle_volume_control.diagnostics.logger import Asynclogger
```

```python
logger = Asynclogger(print_to_console=True)
logger.start()
logger.info("Logger initialized and running.")
logger.warning("This is a warning message.")
logger.error("This is an error message.")
logger.stop()
```
